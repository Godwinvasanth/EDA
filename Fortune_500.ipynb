```python
import pandas as pd
import numpy as np
```

```python
file_path = "/Users/godwinvasanthsamueldhinakar/EDA/F500.csv"

df= pd.read_csv(file_path,sep=';')
print('**************************************************')
print(df.info())
```

```python
df.head(10)
```

```python
df['rev']=pd.to_numeric(df['revenues'].str.replace(',', '').str.replace('$', ''),errors='coerce')
df['prof']=pd.to_numeric(df['profits'].str.replace(',', '').str.replace('$', ''),errors='coerce')
df['asset']=pd.to_numeric(df['assets'].str.replace(',', '').str.replace('$', ''),errors='coerce')
df['marketval']=pd.to_numeric(df['marketvalue'].str.replace(',', '').str.replace('$', ''),errors='coerce')
df.drop(['revenues','profits','assets','marketvalue'],axis=1,inplace=True)
```
