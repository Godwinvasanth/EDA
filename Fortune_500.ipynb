```python
import pandas as pd
import numpy as np
```

```python
file_path = "/Users/godwinvasanthsamueldhinakar/EDA/F500.csv"

df= pd.read_csv(file_path,sep=';')
print('**************************************************')
print(df.info())
```

```python
df.head(10)
```

```python
df['revenues']=pd.to_numeric(df['revenues'].str.replace(',', '').str.replace('$', ''),errors='coerce')
df['profits']=pd.to_numeric(df['profits'].str.replace(',', '').str.replace('$', ''),errors='coerce')
df['assets']=pd.to_numeric(df['assets'].str.replace(',', '').str.replace('$', ''),errors='coerce')
df['marketvalue']=pd.to_numeric(df['marketvalue'].str.replace(',', '').str.replace('$', ''),errors='coerce')
df['employees']=pd.to_numeric(df['employees'].str.replace(',', ''),errors='coerce')
df['previousrank']=df['previousrank'].astype('Int64')
df[['revenuechange', 'profitschange']] = df[['revenuechange', 'profitschange']].apply(
    lambda col: pd.to_numeric(
        col.astype(str).str.replace('%', '', regex=False).replace('-', 'NaN'),
        errors='coerce'
    )
)
# df[['profitschange','assetschange','marketvaluechange']]=df[['profitschange','assetschange','marketvaluechange']].replace('%', '', regex=True).astype('float')
# df.drop(['id','revenues','profits','assets','marketvalue','previousrank','employees','profitschange','revenuechange'],axis=1,inplace=True)
```
